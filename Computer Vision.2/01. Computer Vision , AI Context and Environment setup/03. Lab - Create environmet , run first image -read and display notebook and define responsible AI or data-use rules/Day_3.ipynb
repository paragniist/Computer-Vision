**Define Responsible AI**
Responsible AI is the practice of designing, developing, and deploying artificial intelligence systems in a way that is ethical, safe, fair, and accountable — ensuring AI benefits people without causing unintended harm.
<br>
<br>
**It is built on some pillars:**
1. Fairness — AI shouldn't discriminate or produce biased outcomes against individuals or groups (e.g., a hiring algorithm shouldn't favor one gender or race over another based on flawed training data).

2. Transparency & Explainability — People should be able to understand how and why an AI system made a decision, not just receive an unexplainable output.

3. Accountability — There must be clear ownership for what an AI system does. If it causes harm, someone — a person or organization — is responsible for it, not "the algorithm."

4. Privacy & Data Protection — AI systems must handle personal and sensitive data responsibly, respecting consent and regulations like GDPR.

5. Safety & Reliability — AI should behave predictably and safely, even in edge cases, and shouldn't be easily manipulated (e.g., adversarial attacks) or cause physical/digital harm.

6. Human Oversight — Humans should remain in control of critical decisions, with the ability to intervene, override, or shut down an AI system when needed.
